# Term Structure Models

## Background

Please start by building an n = 10-period binomial model for the short-rate, r (i, j). The lattice parameters are:  r (0 , 0) = 5 %, u = 1.1, d = 0.9 and q = 1 - q = 1/2

```python
# parameters

n = 10
r = 0.05
u = 1.1
d = 0.9
q = 0.5

# ----------------------------------
# short rate lattice
# ----------------------------------
rates = np.zeros((n+1, n+1))
for i in range(n+1):
    for j in range(i+1):
        rates[j,i] = r*(u**j)*(d**(i-j))

rates_df = pd.DataFrame(rates)

rates_df
```

## q1: compute the price of a zero coupon bond that matures at time t and has a face value of $100

```python
# ----------------------------------
# zcb lattice
# ----------------------------------
zcb = np.zeros((n+1, n+1))
FV = 100

#terminal payoff
zcb[:, n] = FV

# backward induction
for i in range(n-1, -1, -1):
    for j in range(i+1):
        expected_value = (
            q * zcb[j+1, i+1]
            + (1-q) * zcb[j, i+1]
        )
        zcb[j, i] = expected_value / (1 + rates[j, i])

zcb_df = pd.DataFrame(zcb)

zcb_df
```
