# Option Pricing in the Multi-Period Binomial Model

You should build a 15-period binomial model whose parameters are calibrated to a Black-Scholes geometric Brownian motion model with:
* T = 0.25 years
* S0 = 100
* r = 2 %
* σ = 30 %
* dividend yield of  c = 1 %

Your binomial model should use a value of  u = 1.0395.

```python
import numpy as np
import pandas as pd
```

```python
# parameters
T = 0.25
S0 = 100
r = 0.02
sigma = 0.30
c = 0.01
N = 15

# Given
u = 1.0395
d = 1/u

dt = T/N

# Risk-neutral probability
q = (np.exp((r-c)*dt)-d)/(u-d)

# Discount factor
disc = np.exp(-r*dt)

# -----------------------
# STOCK TREE
# -----------------------

stock = np.zeros((N+1, N+1))

for i in range(N+1):
    for j in range(i+1):
        stock[j,i] = S0*(u**j)*(d**(i-j))

stock_df = pd.DataFrame(stock)
# =====================================================
# AMERICAN CALL
# =====================================================

K_call = 110

call = np.zeros((N+1, N+1))

# terminal payoffs
for j in range(N+1):
    call[j,N] = max(stock[j,N] - K_call, 0)

# backward induction
for i in range(N-1,-1,-1):
    for j in range(i+1):

        continuation = disc*(
            q*call[j+1,i+1]
            +(1-q)*call[j,i+1]
        )

        exercise = max(stock[j,i]-K_call,0)

        call[j,i] = max(exercise,continuation)

print("American Call Price =", round(call[0,0],4))
call_df = pd.DataFrame(call)
# =====================================================
# AMERICAN PUT
# =====================================================

K_put = 110

put = np.zeros((N+1, N+1))

# terminal payoffs
for j in range(N+1):
    put[j,N] = max(K_put - stock[j,N],0)

# backward induction
for i in range(N-1,-1,-1):
    for j in range(i+1):

        continuation = disc*(
            q*put[j+1,i+1]
            +(1-q)*put[j,i+1]
        )

        exercise = max(K_put-stock[j,i],0)

        if exercise > continuation:
            print(f"American Put: Exercise at node ({j},{i})")

        put[j,i] = max(exercise,continuation)

print("American Put Price =", round(put[0,0],4))
put_df = pd.DataFrame(put)
```

Compute the fair value of an American call option with strike K = 110 and maturity n = 10 periods where the option is written on a futures contract that expires after 15 periods. The futures contract is on the same underlying security of the previous questions. What is also the earliest period you might want to excerise the american futures option?

```python
# Futures tree
futures = np.zeros((N+1,N+1))

for i in range(N+1):
    tau = (N-i)*dt

    for j in range(i+1):
        futures[j,i] = stock[j,i]*np.exp((r-c)*tau)

# American call on futures
M = 10
K = 110

call = np.zeros((N+1,N+1))

# terminal payoff
for j in range(M+1):
    call[j,M] = max(futures[j,M]-K,0)

earliest = None

for i in range(M-1,-1,-1):

    for j in range(i+1):

        continuation = disc*(
            q*call[j+1,i+1]
            +(1-q)*call[j,i+1]
        )

        exercise = max(futures[j,i]-K,0)

        if exercise > continuation and earliest is None:
            print(f"Futures: Exercise at node ({j},{i})")
            earliest = i

        call[j,i] = max(exercise,continuation)

print("American futures call =", call[0,0])
print("Earliest exercise period =", earliest)
```
