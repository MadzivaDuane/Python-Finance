## Basic Data Types

* Python is a dynamically typed language, which means that the Python interpreter infers the type of an object at runtime. In comparison, compiled languages like C are generally statically typed
* Key data types:
    * int
    * float - most needed in finance
    * string - probably the data type python is one of the best at
    * bool

## Basic Data Structures

* Data structures are objects that contain a possibly large number of other objects
* Built in python data structures include:
    * tuple: collection of arbitrary objects; only a few methods available
    * list: collection of arbitrary objects; many methods available
    * dict: key-value store object
    * set: unordered collection object for other unique objects
    * arrays: generally structure other (fundamental) objects in rows and columns

## Numpy Data Structures

* A list can be considered a 1D array
* Lists however, are an inefficient way to make arrays as they are not designed for arrays

```python
v = [1, 2, 3, 4, 5]
m = [v, v, v]

m
```

```python
# indexing in lists
m[0] # first list

m[0][0] # first element in first list
```

A more efficient way to make arrays will be thru np arrays:

```python
a = np.array([1, 2, 3, 4 ,5])

# some operations
a.sum() # sum everything
a.std() # standard deviation
a.cumsum() # cumulative sum

# vectorized mathematical computations
a*2 # here a list returns the array, twice 
a**2 # elemtn wise square

# making new arrays with existing arrays
b = np.array([a, a*2])

# indexing a +1D array
b[0] # first np.array

b.sum() # sums everyting in the array
```
